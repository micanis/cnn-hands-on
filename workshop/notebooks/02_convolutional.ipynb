# 3. 畳み込みとは

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from IPython.display import display
import time
from PIL import ImageOps
import torchvision.transforms as transforms

# 画像を60x60にして白黒に変換
# --- この数値は自由に変えて良い ---
N = 60 # 224以下で設定
img_small = transforms.Resize((N, N))(img)
img_gray = ImageOps.grayscale(img_small)
img_array = np.array(img_gray) / 255.0

# 3x3エッジ抽出フィルタ
v_kernel = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

h_kernel = np.array([
    [-1, -1, -1],
    [0, 0, 0],
    [1, 1, 1]
])

h, w = img_array.shape
output_array = np.zeros((h-2, w-2))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

# 左側（入力画像）の初期設定
ax1.imshow(img_array, cmap='gray')
rect1 = patches.Rectangle((-0.5, -0.5), 3, 3, linewidth=2, edgecolor='red', facecolor='none')
ax1.add_patch(rect1)
ax1.set_title("Input Image & Filter (3x3)")
ax1.axis('off')

# 右側（出力画像）の初期設定
im2 = ax2.imshow(output_array, cmap='gray', vmin=-3, vmax=3)
rect2 = patches.Rectangle((-0.5, -0.5), 1, 1, linewidth=2, edgecolor='red', facecolor='none')
ax2.add_patch(rect2)
ax2.set_title("Feature Map (Output)")
ax2.axis('off')

# グラフを画面に表示
display_handle = display(fig, display_id=True)
plt.close(fig)

time.sleep(1)

skip_steps = 10  # 10歩ごとに画面を更新
step_count = 0

for y in range(h - 2):
    for x in range(w - 2):
        # 1. 画像から 3x3 の領域を切り出して掛け算（畳み込み）
        region = img_array[y:y+3, x:x+3]
        val = np.sum(region * v_kernel) # h_kernelに変えると？？
        output_array[y, x] = val
        step_count += 1

        if step_count % skip_steps == 0 or (y == h-3 and x == w-3):
            rect1.set_xy((x-0.5, y-0.5))
            im2.set_data(output_array)
            rect2.set_xy((x-0.5, y-0.5))

            display_handle.update(fig)
            time.sleep(0.05) # アニメーションのスピード調整
```

# 4. 畳み込みの実装

```python
###
# 演習: 畳み込みの手動実装
###

kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

def conv2d(img, kernel, padding=0):
    if padding > 0:
        img = np.pad(img, padding, mode='constant')
    # h (height:高さ), w (width:幅)を画像のshapeから取得するには？
    h, w = img.shape[:?]

    # kh (kernel height: カーネルの高さ), kw (kernel width: カーネルの幅)を取得するには？
    # img (画像)も kernel も同じ配列である
    kh, kw = kernel.?

    # 出力の高さを、入力画像の高さに合わせる
    out_h = h - kh + ?

    # 出力の幅を、入力画像の幅に合わせる
    out_w = w - kw + ?

    # 出力の高さ・幅で配列を初期化する
    out = np.?((out_h, out_w))

    for i in range(out_h):
        for j in range(out_w):

            # 画像の一部とカーネルの掛け算をしたい
            # うまく、スライスを使って画像の一部を抜き出す
            region = img[?:?+?, ?:?+?]

            # 抜き出した画像とカーネルの掛け算の合計値を、出力の配列に格納する
            out[i, j] = np.sum(region * kernel)
    return out
```

```python
# 畳み込み実行
edge = conv2d(np.array(img_gray), kernel, padding=1)

# 可視化
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title('Input image')
axes[1].imshow(np.abs(edge), cmap='gray')
axes[1].set_title('Result')
plt.show()
```
