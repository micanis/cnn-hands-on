```python
###
# セットアップ
###

from pathlib import Path
import sys

try:
    from google.colab import drive
    drive.mount("/content/drive")
    ROOT_PATH = Path("/content/drive/MyDrive/cnn-hands-on")
except Exception:
    ROOT_PATH = Path.cwd()

if str(ROOT_PATH) not in sys.path:
    sys.path.append(str(ROOT_PATH))

print("ROOT_PATH:", ROOT_PATH)
```

```python
###
# Cell 01
# 学習済みYOLOv8nを読み込む
###

from IPython.display import Image, display
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
print(model.names)
```

```python
###
# Cell 02
# サンプル画像で一般物体検出を試す
###

sample_image = "https://ultralytics.com/images/bus.jpg"

results = model.predict(
    source=sample_image,
    conf=0.25,
    save=True,
)

save_dir = Path(results[0].save_dir)
display(Image(filename=str(save_dir / "bus.jpg")))
```

```python
###
# Cell 03
# Webカメラで画像を撮影する
###

from utils.camera import take_photo

photo_path = take_photo("camera.jpg")
display(Image(filename=photo_path))
```

```python
###
# Cell 04
# 撮影した画像で推論する
###

results = model.predict(source="camera.jpg", conf=0.25, save=True)
output_path = Path(results[0].save_dir) / "camera.jpg"
display(Image(filename=str(output_path)))
```
